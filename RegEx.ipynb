re must be imported.

```python
import re
```

A sample string

```python
tr = 'an example word:cat!!'
```

Use re.search with the first argument being the regex and the second the string.

```python
match = re.search(r'word:\w\w\w',tr)
```

It returns a match object. Note that \w matches any alphanumeric character

Note the raw string. What happens if you change it to a regular string

```python
match
```

```python
match.group(0)
```

```python
match.group()
```

```python
if match:
    print('found', match.group())
else:
    print('did not find')
```

```python
def check(match):
    if match:
        print('found', match.group())
    else:
      print('did not find')
```

```python
check(match)
```

```python
def find(pat, text):
  match = re.search(pat, text)
  if match:
    print(match.group())
  else:
    print('not found')
```

```python
find (r':\w\w\w', 'blah :cat blah blah')
```

Finding multiple instances of pattern. findall

```python
str2 = 'purple alice@google.com, blah monkey bob@abc.com blah dishwasher'
```

```python
pat
```

```python
pat0 = r'[\w.-]+@[\w.-]+'
```

```python
match = re.search(pat0, str2)
```

```python
match.group()
```

```python
emails = re.findall(pat0,str2)
```

```python
emails
```

```python
str3 = 'My 2 favorite numbers are 19 and 42'
```

```python
nums = re.findall('[0-9]+', str3)
```

```python
nums
```
